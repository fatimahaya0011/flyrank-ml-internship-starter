## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The sample has 30,000 pages and 44 columns. 30 of the columns are numbers. The column with the most missing data is provider_used, missing in 71.5% of rows. The data also has columns like search_volume, cpc, competition and word_count, so I can look at what makes a page worth refreshing first.

```python
import pandas as pd

URL = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(URL)

missing = df.isna().mean()
print("Rows:", len(df))
print("Columns:", df.shape[1])
print("Number columns:", df.select_dtypes("number").shape[1])
print("Most missing column:", missing.idxmax(), "-", round(missing.max() * 100, 1), "% missing")
print(df.columns.tolist())
```

 54.2% of pages have a "down" trend, and the middle page was last updated 20 days ago.

```python
print(df["trend_direction"].value_counts(normalize=True).round(3))
print("Median days since last update:", df["days_since_last_update"].median())
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can say: my results are observed in this sample, measured, directional, and meant for decision-support.
What I will never say: that I proved a cause, or that I predicted Google's ranking.

```python
claim = "My results are observed in this sample and are directional decision-support."
banned = ["predict google", "prove", "proves", "caused", "causes", "guarantee"]
print("Risky words found:", [w for w in banned if w in claim.lower()])
```
